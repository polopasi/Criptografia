## Coste de las operaciones básicas

$n$ y $m$ enteros, $n>m$.

<span style="color: purple">
Coste suma $n+m$:   
    
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $\mathcal{O}(\log{}n)$
</span>

<br>

<span style="color:purple">
Coste producto $n\cdot m$: 
    
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;$\mathcal{O}(\log^2 n)$;  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Karatsuba $\mathcal{O}((\log{}n)^{\log _{2}3})$;   
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Schönhage-Strassen $\mathcal{O}(\log{}n\, \log\log{}n\, \log\log\log{}n)$;  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Integer multiplication in time $\mathcal{O}(\log{}n\, \log\log{}n)$ D. Harvey and J. van der Hoeven, 	Annals of Mathematics, March 2021.
</span>

<br>

Karatsuba:  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; Si $n=a 2^{t/2}+b$ y $m=c 2^{t/2}+d$ entonces  
$$n\cdot m=ac2^t+(bc+ad)2^{t/2}+bd=ac2^t+(ac+bd-(a-b)(c-d))2^{t/2}+bd$$

## Divisibilidad 
    
**Definición:**  
<span style="color:teal"> 
Diremos que $a\neq 0$ divide a $b$ si existe $q$ tal que $b=q\,a$.   
Si $a$ divide a $b$ escribiremos $a\vert b$.
</span>
 
<br>


**Teorema división euclidiana:**  
<span style="color:blue"> 
Sean $a, b\in\mathbb Z$, $b\geq 1$, entonces existen $q, r\in\mathbb Z$ tales que
	$$
	a=q\cdot b+r \quad 0\leq r< b.
	$$
	$q$ (cociente) y $r$ (resto) son únicos.
 </span>
 
<br>

**Máximo común divisor:**   
<span style="color:teal"> 
$d> 0$ es el máximo común divisor de $a$ y $b\neq0$ si:  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; i) $d\vert a$, $d\vert b$,  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; ii) $\forall c$ tal que  $c\vert a$ y $c\vert b$ entonces $c\vert d$.  
Escribiremos $d=(a,b)=mcd(a,b)=gcd(a,b)$
</span>
 
<br>

### Algoritmo extendido de Euclides

**Identidad de Bezout:**   
<span style="color:blue"> 
 Dados $a$ y $b$  existen $x$ e $y$ tales que
	$
	ax+by=mcd(a,b)
	$
 </span>
 
<br>

$a>b>0$  
**exGCG(a,b)**  
&nbsp;&nbsp;&nbsp;&nbsp; $r_0 = a$, $r_1 = a$  
&nbsp;&nbsp;&nbsp;&nbsp; $x_0 = 1$, $x_1 = 0$  
&nbsp;&nbsp;&nbsp;&nbsp; $y_0 = 0$, $y_1 = 1$  
&nbsp;&nbsp;&nbsp;&nbsp; $k = 1$  
&nbsp;&nbsp;&nbsp;&nbsp; **while** $r_k\neq 0$  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $r_{k-1}=q_k\, r_k + r_{k+1}$,&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $0\leq r_{k+1}<r_k$  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $x_{k+1}=x_{k-1}-q_k\, x_k$  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $y_{k+1}=y_{k-1}-q_k\, y_k$  
&nbsp;&nbsp;&nbsp;&nbsp; **end while**  
&nbsp;&nbsp;&nbsp;&nbsp; **return**  $(a,b)=r_{k-1}$, $x=x_{k-1}$, $y=y_{k-1}$ `último resto no nulo`


<span style="color: purple">
Coste: $\mathcal{O}(\log^2{}n)$
</span>

### Función $\varphi$ de Euler

**Definición:** <span style="color:teal"> 
$a$ y $b$ son relativamente primos o coprimos si $mcd(a,b)$=1.
</span>
 
<br>

**Definición:** <span style="color:teal"> 
Sea $n>0$, denotaremos por $\varphi(n)$ el número de
enteros positivos menores o iguales que $n$ y que son relativamente primos con $n$:
$$
\varphi(n)=\left\vert\left\{1\leq b\leq n : mcd(b,n)=1\right\}\right\vert
$$
</span>
 
<br>

**Propiedades:**   
<span style="color:blue">
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; i) $\varphi(1)=1$.  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; ii) Si $p$ primo, &nbsp;&nbsp; $\varphi(p)=p-1$.  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; iii) Si $p$ primo, &nbsp;&nbsp; $\varphi(p^\alpha)=p^{\alpha-1}(p-1)$.  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; iv) Si $mcd(a,b)=1$, &nbsp;&nbsp;$\varphi(a\,b)=\varphi(a)\,\varphi(b)$.  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; v) Si $n=\prod p_i^{\alpha_i}$, &nbsp;&nbsp;
      $\varphi(n)=\prod p_i^{\alpha_i-1}(p_i-1)=
       n\prod\left(1-\frac{1}{p_i}\right)$.  
</span>
 
<br>

## Aritmética modular 
   
**Definición:** <span style="color:teal"> Sea $n\geq 1$, $a$ es congruente
con $b$ módulo $n$  si $n\vert(a-b)$.  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; Escribiremos $a\equiv b\mod n$.
 </span>
 
<br>


**Definición:** <span style="color:teal">Un conjunto de números enteros tales que cualquier entero es congruente módulo $n$ con exactamente un elemento del conjunto recibe el nombre de conjunto completo de residuos módulo $n$.
 </span>
 
<br>


**Definición:** <span style="color:teal">$\mathbb Z_n$: cualquier conjunto completo de residuos módulo $n$.
$$
	\mathbb Z_n=\left\{0,1,\ldots,n-2,n-1\right\}=
	\left\{1,2,\ldots,n-1,n\right\}.
$$
 </span>
 
<br>


**Definición:** <span style="color:teal"> $a\in\mathbb Z_n$ es invertible si
      $\exists b\in\mathbb Z_n$ tal que $a\,b\equiv 1\mod n$.
 </span>
 
<br>


**Teorema:** <span style="color:blue"> 
$a\in\mathbb Z_n$ es invertible $\Leftrightarrow$ $mcd(a,n)=1$.
 </span>
 
<br>

<span style="color:purple"> 
	Coste de cálculo de inversos módulo n usando el algoritmo extendido de Euclides: $\mathcal{O}(\log^2{}n)$
 </span>
 
<br>
<br>

**Definición:** <span style="color:teal">  $\mathbb Z_n^*$: Conjunto de los elementos
invertibles de $\mathbb Z_n$.
 </span>
 
<br>

Notemos que $\left\vert\mathbb Z_n^*\right\vert=\varphi(n)$.

### Pequeño teorema de Fermat

**Pequeño teorema de Fermat:**  
<span style="color:blue"> 
Sea $p$ primo, entonces:  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; i) $\forall a$ con $mcd(p,a)=1$, &nbsp;&nbsp;&nbsp;&nbsp; $$a^{p-1}\equiv 1\mod p.$$  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; ii) $\forall a$, &nbsp;&nbsp;&nbsp;&nbsp; $$a^{p}\equiv a \mod p.$$
 </span>
 
<br>


**Corolario:** 
<span style="color:blue">
Si $mcd(p,a)=1$, $p$ primo y $r\equiv s\mod p-1$ entonces
$$
a^{r}\equiv a^{s}\mod p.
$$
 </span>
 
<br>

### Teorema de Euler

<br>

**Teorema de Euler:** 
<span style="color:blue"> 
Si $mcd(n,a)=1$ entonces
$$
a^{\varphi(n)}\equiv 1\mod n.
$$
 </span>
 
<br>

**Corolario:** <span style="color:blue">
Si $(n,a)=1$ y $r\equiv s\mod \varphi(n)$ entonces
$$
a^{r}\equiv a^{s}\mod n.
$$
 </span>

En la exponenciación modular podemos suponer que el tamaño del exponente es menor que $n$.

 
<br>

**Corolario:** <span style="color:blue">
Si $mcd(n,a)=1$ entonces
$$
a^{-1}\equiv a^{\varphi(n)-1}\mod n.
$$
 </span>

### Exponenciación modular $a^{e}\mod n$

Sea $e=e_r2^r+\cdots+e_12+e_0$, $e_i\in\{0,1\}$, $e_r=1$

**PowerMod(a, e, n)**  
&nbsp;&nbsp;&nbsp;&nbsp; $y\gets 1$, $i\gets r$  
&nbsp;&nbsp;&nbsp;&nbsp; **while** $i \geq 0$  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $y\gets y^2 \mod n$ &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;  `▹No es lo mismo elevar al cuadrado que multiplicar`  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; **if** $e_i=1$}  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;			$y\gets y\cdot a \mod n$ `▹No es lo mismo elevar al cuadrado que multiplicar`  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; **end if**  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $i \gets i-1$  
&nbsp;&nbsp;&nbsp;&nbsp; **end while**  
&nbsp;&nbsp;&nbsp;&nbsp; **return** $y$  



<br>

<span style="color:purple">
	Coste de calcular $a^{e}\mod n$: $\mathcal{O}(\log^3{}n)$
</span>

<br>
<br>
<span style="color:red">
	Coste de calcular $a^{-1}\equiv a^{\varphi(n)-1}\mod n$: $\mathcal{O}(\log^3{}n)$
</span>

<br>
<br>

### Teorema chino de los restos

<p>
<span style="color:blue">
Sean $n_1,\ldots, n_k\geq 1$ tales que $(n_i,n_j)=1,\ i\neq j$.
El sistema
$$
\begin{array}{c}
x\equiv a_1 \mod n_1\\
\vdots\\
x\equiv a_k \mod n_k
\end{array}
$$
tiene una única solución módulo $n=n_1\,n_2\ldots n_k$.
</span>
</p>


<br><br><br>

<p>
<span style="color:black">
La solución es:
$$
\sum_i M_i L_i a_i \mod n
$$
siendo $L_i=\frac{n}{n_i}$ y $M_i$ el inverso de $L_i$ módulo $n_i$.
</span>
</p>

<br>
<span style="color:purple">
	Coste: $\mathcal{O}(\log^2{}n)$
</span>
